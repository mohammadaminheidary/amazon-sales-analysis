### 1. What was the total sales amount?

```python
total_revenue = df["Amount"].sum()
total_orders = df["Order ID"].nunique()
total_quantity = df["Qty"].sum()

sales_summary = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Orders",
        "Total Quantity Sold"
    ],
    "Value": [
        f"{total_revenue:,.2f}",
        f"{total_orders:,}",
        f"{total_quantity:,}"
    ]
}, index=range(1, 4))


sales_summary
```

### 2. How has sales changed over time?

```python
monthly_sales = (
    df.groupby(df["Date"].dt.to_period("M"), observed=True)
    .agg(
        Revenue=("Amount", "sum"),
        Quantity_Sold=("Qty", "sum")
    )
    .reset_index()
)

monthly_sales["Month"] = monthly_sales["Date"].astype(str)

monthly_sales["Revenue_Change"] = monthly_sales["Revenue"].diff()

monthly_sales["Revenue_Change_%"] = (
    monthly_sales["Revenue"].pct_change() * 100
)

monthly_sales["Sales_Trend"] = monthly_sales["Revenue_Change"].apply(
    lambda x: (
        "Increase" if x > 0
        else "Decrease" if x < 0
        else "No Change"
    )
)

monthly_sales = monthly_sales[
    [
        "Month",
        "Revenue",
        "Quantity_Sold",
        "Revenue_Change",
        "Revenue_Change_%",
        "Sales_Trend"
    ]
]

monthly_sales["Revenue"] = monthly_sales["Revenue"].map(
    lambda x: f"{x:,.2f}"
)

monthly_sales["Quantity_Sold"] = monthly_sales["Quantity_Sold"].map(
    lambda x: f"{x:,}"
)

monthly_sales["Revenue_Change"] = monthly_sales["Revenue_Change"].map(
    lambda x: f"{x:,.2f}" if pd.notna(x) else "-"
)

monthly_sales["Revenue_Change_%"] = monthly_sales["Revenue_Change_%"].map(
    lambda x: f"{x:,.2f}%" if pd.notna(x) else "-"
)

monthly_sales.index = range(1, len(monthly_sales) + 1)

monthly_sales
```

### 3. What is the average order value?

```python
total_revenue = df["Amount"].sum()
total_orders = df["Order ID"].nunique()

average_order_value = total_revenue / total_orders

aov_summary = pd.DataFrame({
    "Metric": [
        "Total Revenue",
        "Total Orders",
        "Average Order Value"
    ],
    "Value": [
        f"{total_revenue:,.2f}",
        f"{total_orders:,}",
        f"{average_order_value:,.2f}"
    ]
}, index=range(1, 4))

aov_summary
```

### 4. Which days recorded the highest sales?

```python
top_sales_days = (
    df.groupby(df["Date"].dt.date)
    .agg(
        Revenue=("Amount", "sum"),
        Quantity_Sold=("Qty", "sum"),
        Orders=("Order ID", "nunique")
    )
    .reset_index()
    .rename(columns={"Date": "Day"})
    .sort_values("Revenue", ascending=False)
    .head(15)
)

top_sales_days["Revenue"] = top_sales_days["Revenue"].map(
    lambda x: f"{x:,.2f}"
)

top_sales_days["Quantity_Sold"] = top_sales_days["Quantity_Sold"].map(
    lambda x: f"{x:,}"
)

top_sales_days["Orders"] = top_sales_days["Orders"].map(
    lambda x: f"{x:,}"
)

top_sales_days.index = range(1, len(top_sales_days) + 1)

top_sales_days
```
