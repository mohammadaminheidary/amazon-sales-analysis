# Cleaning🧹

```python
shape_before = df.shape
print(f'Number of rows and columns before cleaning: {df.shape}')
```

## Cleaning missing values in column Courier Status

```python
df['Courier Status'].unique()
```

```python
df.loc[df["Courier Status"].isna(), "Status"].value_counts()
```

```python
df.loc[
    (df["Courier Status"].isna()) & 
    (df["Status"] == "Cancelled"),
    "Courier Status"
] = "Cancelled"
```

```python
df.shape
```

```python
missing_courier_rows = df[
    (df["Courier Status"].isna()) &
    (df["Status"] != "Cancelled")
]
missing_courier_rows.shape
```

```python
df = df.drop(missing_courier_rows.index)
```

```python
df.shape
```

- Delete 11 rows from the data

## Cleaning missing values in columns currency & Amount

```python
df["currency"] = df["currency"].fillna("unknown")
df["Amount"] = df["Amount"].fillna("unpaid")
```

```python
print(f'Number of NaN values in column currency: {df['currency'].isna().sum()}')
print(f'Number of NaN values in column Amount: {df['Amount'].isna().sum()}')
```

## Cleaning missing values in columns ship-city & ship-state & ship-postal-code & ship-country

```python
print(f'Number of NaNs before cleaning: {df["ship-city"].isna().sum()}')
```

```python
df = df.dropna(subset=["ship-city"])
```

```python
print(f'Number of NaNs after cleaning: {df["ship-city"].isna().sum()}')
```

## Cleaning missing values in column promotion-ids

```python
print(f'Number of NaNs before cleaning: {df["promotion-ids"].isna().sum()}')
```

```python
df["promotion-ids"] = df["promotion-ids"].fillna("No Promotion")
```

```python
print(f'Number of NaNs after cleaning: {df["promotion-ids"].isna().sum()}')
```

## Cleaning missing values in columns fulfilled-by & Unnamed: 22

```python
print(f'Number of NaNs before cleaning: {df.shape}')
```

```python
df = df.drop(columns=["fulfilled-by", "Unnamed: 22"])
```

```python
print(f'Number of NaNs after cleaning: {df.shape}')
```

```python
shape_after = df.shape
print(f'Number of rows and columns after cleaning: {df.shape}')
```

```python
print(f'Number of rows and columns deleted from the database: {[shape_before[0] - shape_after[0], shape_before[1] - shape_after[1]]}')
```
