# Dataset Columns Description

این دیتاست شامل اطلاعات سفارش‌های فروش آمازون است که جزئیات مربوط به سفارش، محصول، وضعیت ارسال، اطلاعات مشتری، مبلغ فروش و نوع تکمیل سفارش را نگهداری می‌کند.

در ادامه توضیح کامل هر ستون، نوع داده احتمالی و مقادیر قابل قبول آن آورده شده است.

| Column | Description | Possible Values / Examples |
|---|---|---|
| `index` | شماره ردیف ایجاد شده هنگام ذخیره یا استخراج داده‌ها. این ستون معمولاً یک شناسه داخلی برای هر رکورد است و بخشی از اطلاعات اصلی سفارش محسوب نمی‌شود. | عدد ترتیبی مانند `0`, `1`, `2`, ... |
| `Order ID` | شناسه یکتای سفارش در آمازون. هر سفارش با یک کد مشخص ثبت می‌شود و برای پیگیری سفارش استفاده می‌شود. | رشته متنی مانند `408-1234567-1234567` |
| `Date` | تاریخ ثبت سفارش توسط مشتری. این ستون برای تحلیل روند فروش در طول زمان استفاده می‌شود. | تاریخ‌هایی مانند `04-30-2022`, `2022-05-01` |
| `Status` | وضعیت فعلی سفارش در چرخه فروش. نشان می‌دهد سفارش تکمیل شده، لغو شده یا در چه مرحله‌ای قرار دارد. | `Shipped`, `Cancelled`, `Pending`, `Delivered`, `Returned` |
| `Fulfilment` | روش انجام و پردازش سفارش. مشخص می‌کند چه کسی مسئول آماده‌سازی و ارسال سفارش بوده است. | `Amazon`, `Merchant` |
| `Sales Channel` | کانال فروش که سفارش از طریق آن ثبت شده است. | `Amazon.in`, `Amazon.com` یا سایر کانال‌های فروش |
| `ship-service-level` | سطح سرویس حمل‌ونقل انتخاب شده برای ارسال سفارش. | `Standard`, `Expedited`, `Priority` |
| `Style` | نوع یا مدل ظاهری محصول. معمولاً برای دسته‌بندی محصولات پوشاک استفاده می‌شود. | مقادیری مانند `Kurta`, `Top`, `Set`, `Dress` |
| `SKU` | شناسه داخلی محصول (Stock Keeping Unit). برای مدیریت موجودی و تشخیص دقیق کالا استفاده می‌شود. | کدهایی مانند `JNE1234`, `SKU-001` |
| `Category` | دسته‌بندی اصلی محصول. | `Set`, `Kurta`, `Western Dress`, `Top`, `Ethnic Dress` |
| `Size` | اندازه یا سایز محصول. بیشتر برای محصولات پوشیدنی استفاده می‌شود. | `XS`, `S`, `M`, `L`, `XL`, `XXL`, `Free Size` |
| `ASIN` | شناسه استاندارد محصول در آمازون (Amazon Standard Identification Number). هر محصول در آمازون یک ASIN دارد. | کدی ۱۰ کاراکتری مانند `B08XYZ1234` |
| `Courier Status` | وضعیت ارسال توسط شرکت حمل‌ونقل. | `Shipped`, `Delivered`, `Cancelled`, `Unshipped`, `Pending` |
| `Qty` | تعداد واحدهای خریداری‌شده از محصول در سفارش. | اعداد صحیح مانند `1`, `2`, `3`, ... |
| `currency` | واحد پول مورد استفاده برای مبلغ سفارش. | `INR`, `USD`, `EUR` |
| `Amount` | مبلغ کل فروش سفارش. این مقدار نشان‌دهنده درآمد حاصل از فروش محصول است. | اعداد مانند `499`, `1299.50`, `2500` |
| `ship-city` | شهر مقصد ارسال سفارش (محل دریافت مشتری). | نام شهرها مانند `Mumbai`, `Delhi`, `Bangalore` |
| `ship-state` | ایالت یا منطقه مقصد ارسال سفارش. | مقادیری مانند `Maharashtra`, `Delhi`, `Karnataka` |
| `ship-postal-code` | کد پستی مقصد ارسال سفارش. | اعداد مانند `400001`, `110001` |
| `ship-country` | کشور مقصد ارسال سفارش. | معمولاً `IN` یا `India` |
| `promotion-ids` | شناسه تخفیف‌ها یا کمپین‌های تبلیغاتی اعمال‌شده روی سفارش. | شناسه‌هایی مانند `Amazon PLCC`, `INRETAIL`, یا مقدار خالی |
| `B2B` | مشخص می‌کند سفارش مربوط به مشتری تجاری (Business-to-Business) است یا مشتری عادی. | `True`, `False` |
| `fulfilled-by` | مشخص می‌کند چه شخص یا سازمانی سفارش را تکمیل کرده است. | `Amazon`, `Easy Ship`, `Merchant` |
| `Unnamed: 22` | ستون اضافی ایجاد شده هنگام ذخیره فایل CSV یا Excel که معمولاً فاقد اطلاعات مفید است. | مقادیر خالی (`NaN`) یا داده‌های اضافی |

# Business Questions & Analysis Expectations

در یک پروژه Data Analysis مربوط به فروش آمازون، هدف اصلی فقط بررسی اعداد نیست؛ بلکه باید بتوانیم از داده‌ها **بینش‌های قابل استفاده برای تصمیم‌گیری کسب‌وکار** استخراج کنیم.

با استفاده از این دیتاست می‌توان سؤالات مختلفی در حوزه‌های فروش، مشتری، محصول، عملیات ارسال و عملکرد کسب‌وکار مطرح کرد.

---

# 1. Sales Performance Analysis (تحلیل عملکرد فروش)

## سوالات قابل بررسی:

### 1. میزان کل فروش چقدر بوده است؟

**هدف تحلیل:**

محاسبه درآمد کلی کسب‌وکار در بازه زمانی موجود.

Metrics:

- Total Revenue
- Total Orders
- Total Quantity Sold

---

### 2. روند فروش در طول زمان چگونه تغییر کرده است؟

سؤالات:

- فروش در چه ماه‌هایی بیشتر بوده است؟
- آیا روند فروش صعودی بوده یا نزولی؟
- چه دوره‌هایی افت فروش داشته‌اند؟

ستون‌های مورد استفاده:

- Date
- Amount
- Qty

Expected Output:

- Monthly Sales Trend
- Seasonal Sales Pattern

---

### 3. میانگین ارزش هر سفارش چقدر است؟

فرمول:
‍‍‍‍```Average Order Value = Total Revenue / Number of Orders```

هدف:

بررسی کیفیت فروش و ارزش هر مشتری.

---

### 4. چه روزهایی بیشترین فروش ثبت شده است؟

تحلیل:

- روزهای پرفروش
- روزهای کم‌فروش

کاربرد:

بهینه‌سازی کمپین‌های تبلیغاتی.

---

# 2. Product Analysis (تحلیل محصولات)

## سوالات قابل بررسی:

---

## 2. کدام دسته محصول بیشترین درآمد را ایجاد کرده است؟

- محصولی فروش کمتر ولی سود بیشتری داشته باشد.

ستون:
```Size```

تحلیل:

- کدام سایز بیشترین تقاضا را دارد؟

## 4. آیا بعضی محصولات عملکرد ضعیفی دارند؟

بررسی:

- محصولات کم‌فروش
- محصولات بدون تقاضا

تصمیم برای:

- حذف محصول
